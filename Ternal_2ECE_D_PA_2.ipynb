**A. REPRODUCIBLE NORMALIZATION PROBLEM**


```python
import numpy as np
```

```python
np.random.seed(2112)
```

```python
x = np.random.randint(10, 101, size=(5, 5))
x
```

```python
X_mean = x.mean()
X_mean
```

```python
X_std = x.std()
X_std
```

```python
X_normalized = (x- X_mean)/X_std
X_normalized
```

```python
print ("Original Array (x):", x)
print ("X_normalized:", X_normalized)
print ("X_normalized mean:", X_normalized.mean())
print ("X_normalized standard deviation:", X_normalized.std())
```

```python
np.save("X_normalized.npy", X_normalized)
```

**B. CUBES DIVISIBLE BY 4 PROBLEM**

```python
c = (np.arange(1, 101)**3).reshape(10,10)
c
```

```python
div_by_4 = c[c % 4 == 0]
div_by_4
```

```python
print ("Shape of C:", c.shape)
print ("Array div_by_4:", div_by_4)
print ("Number of selected elements:", len(div_by_4))
```

```python
np.save("div_by_4", div_by_4)
```

**C. ABOVE-MEAN SQUARES PROBLEM**

```python
S = (np.arange(1, 37)**2).reshape(6,6)
S
```

```python
S_mean = S.mean()
S_mean
```

```python
above_mean = S[S > S_mean]
above_mean
```

```python
print ("S:", S)
print ("S_mean:", S_mean)
print ("S above mean:", above_mean)
print ("Number of selected elements:", len(above_mean))
```

```python
np.save("above_mean.npy", above_mean)
```
